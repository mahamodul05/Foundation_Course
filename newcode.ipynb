```python
def find_max(numbers):
    max_num = numbers[0]
    for num in numbers:
        if num > max_num:
            max_num = num
    return max_num


print(find_max([3, 7, 2, 9, 5]))
```

```python
def count_even(numbers):
    count = 0
    for num in numbers:
        if num % 2 == 0:
            count += 1
    return count

print(count_even([1, 2, 3, 4, 5, 6]))  
```

```python
def reverse_list(numbers):
    reversed_list = []
    for i in range(len(numbers) - 1, -1, -1):
        reversed_list.append(numbers[i])
    return reversed_list

print(reverse_list([1, 2, 3, 4]))
```

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)

print(factorial(5))
```

```python
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
import numpy as np

class LabelClass:
    def __init__(self, cities):
        self.cities = cities
        self.label_encoder = LabelEncoder()
        self.onehot_encoder = OneHotEncoder(sparse_output=False)
        
    def label_encode(self):
        labels = self.label_encoder.fit_transform(self.cities)
        return labels
    
    def one_hot_encode(self):
        labels = self.label_encoder.fit_transform(self.cities)
        labels_reshaped = labels.reshape(-1, 1)
        onehot = self.onehot_encoder.fit_transform(labels_reshaped)
        return onehot
cities = ["paris", "paris", "tokyo", "amsterdam"]
labeler = LabelClass(cities)

print("Label Encoded:", labeler.label_encode())
print("One Hot Encoded:\n", labeler.one_hot_encode())
```
